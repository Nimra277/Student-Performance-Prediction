```python
# STEP 2: DATA PRE-PROCESSING
# Roadmap: NumPy and Pandas

import pandas as pd
import numpy as np

# 1. Dataset overview
print("Dataset Shape:", df.shape)

# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 3. Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# 4. Remove duplicate records
df = df.drop_duplicates().copy()

# 5. Handle missing numerical values
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# 6. Handle missing categorical values
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# 7. Convert categorical column into numeric
df["placement_status"] = (
    df["placement_status"].astype(str).str.strip().str.lower()
    .map({"placed": 1, "not placed": 0})
)

# 8. Display final results
print("\nData Pre-processing Completed!")
print("\nUpdated Dataset Shape:", df.shape)
print("\nUpdated Data Types:")
print(df.dtypes)

display(df.head())
```

```python
# ============================================================
# STUDENT PERFORMANCE PREDICTION AND ANALYSIS SYSTEM
# STEP 3: COMPLETE MATHEMATICS AND STATISTICS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from scipy.stats import norm, binom, shapiro
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
from sklearn.decomposition import PCA

# ------------------------------------------------------------
# 1. DATA VALIDATION
# ------------------------------------------------------------

print("=" * 65)
print("STEP 3: MATHEMATICS AND STATISTICS")
print("=" * 65)

required_columns = [
    "study_hours", "attendance", "sleep_hours",
    "internet_usage", "assignments_completed",
    "previous_score", "exam_score", "placement_status"
]

missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

numeric_columns = [
    "study_hours", "attendance", "sleep_hours",
    "internet_usage", "assignments_completed",
    "previous_score", "exam_score"
]

data = df.copy()

for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors="coerce")

data = data.dropna(subset=numeric_columns + ["placement_status"])

# Placement status must be binary: 0 = Not Placed, 1 = Placed
if data["placement_status"].dtype == "object":
    data["placement_status"] = (
        data["placement_status"].astype(str).str.strip().str.lower()
        .map({"placed": 1, "not placed": 0})
    )

data = data.dropna(subset=["placement_status"])
data["placement_status"] = data["placement_status"].astype(int)

if not set(data["placement_status"].unique()).issubset({0, 1}):
    raise ValueError("placement_status must contain only 0 and 1.")

print("\nDataset shape:", data.shape)
print("Data validation completed.")

# ------------------------------------------------------------
# 2. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n2. DESCRIPTIVE STATISTICS")

descriptive_stats = data[numeric_columns].agg(
    ["mean", "median", "std", "var", "min", "max"]
).T

descriptive_stats["mode"] = data[numeric_columns].mode().iloc[0]
descriptive_stats["range"] = (
    data[numeric_columns].max() - data[numeric_columns].min()
)

print(descriptive_stats.round(3))

# ------------------------------------------------------------
# 3. PROBABILITY OF PLACEMENT
# ------------------------------------------------------------

print("\n3. PROBABILITY OF PLACEMENT")

total_students = len(data)
placed = (data["placement_status"] == 1).sum()
not_placed = (data["placement_status"] == 0).sum()

p_placed = placed / total_students
p_not_placed = not_placed / total_students

print("Total students:", total_students)
print("Placed students:", placed)
print("Not placed students:", not_placed)
print("P(Placed):", round(p_placed, 4))
print("P(Not Placed):", round(p_not_placed, 4))

# ------------------------------------------------------------
# 4. CONDITIONAL PROBABILITY
# ------------------------------------------------------------

print("\n4. CONDITIONAL PROBABILITY")

attendance_threshold = data["attendance"].median()

high_attendance = data[data["attendance"] >= attendance_threshold]

if len(high_attendance) > 0:
    conditional_probability = (
        high_attendance["placement_status"] == 1
    ).mean()
else:
    conditional_probability = np.nan

print("Attendance threshold:", attendance_threshold)
print("P(Placed | Attendance >= Median):",
      round(conditional_probability, 4))

# ------------------------------------------------------------
# 5. CONFIDENCE INTERVAL FOR EXAM SCORE
# ------------------------------------------------------------

print("\n5. 95% CONFIDENCE INTERVAL")

exam_scores = data["exam_score"]
sample_mean = exam_scores.mean()
sample_std = exam_scores.std(ddof=1)
n = len(exam_scores)

standard_error = sample_std / np.sqrt(n)
confidence_level = 0.95

t_critical = stats.t.ppf(
    (1 + confidence_level) / 2, df=n - 1
)

margin_error = t_critical * standard_error

lower_ci = sample_mean - margin_error
upper_ci = sample_mean + margin_error

print("Mean Exam Score:", round(sample_mean, 3))
print("Standard Error:", round(standard_error, 3))
print("95% Confidence Interval:",
      (round(lower_ci, 3), round(upper_ci, 3)))

# ------------------------------------------------------------
# 6. CORRELATION ANALYSIS
# ------------------------------------------------------------

print("\n6. CORRELATION MATRIX")

correlation_matrix = data[numeric_columns].corr()

print(correlation_matrix.round(3))

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Student Performance Correlation Matrix")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 7. CONTINUOUS PROBABILITY DISTRIBUTION
# Exam Score: Normal Distribution Analysis
# ------------------------------------------------------------

print("\n7. CONTINUOUS PROBABILITY DISTRIBUTION")

mu = exam_scores.mean()
sigma = exam_scores.std(ddof=1)

x = np.linspace(exam_scores.min(), exam_scores.max(), 300)
pdf = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 5))
sns.histplot(
    exam_scores,
    bins=30,
    stat="density",
    color="skyblue",
    edgecolor="black"
)
plt.plot(x, pdf, linewidth=2, label="Normal Distribution")
plt.axvline(mu, linestyle="--", label="Mean")
plt.title("Exam Score Normal Distribution")
plt.xlabel("Exam Score")
plt.ylabel("Density")
plt.legend()
plt.tight_layout()
plt.show()

print("Mean:", round(mu, 3))
print("Standard Deviation:", round(sigma, 3))

# Shapiro-Wilk test uses a sample of up to 5000 observations
normality_sample = exam_scores.sample(
    min(5000, len(exam_scores)),
    random_state=42
)

shapiro_stat, shapiro_p = shapiro(normality_sample)

print("Shapiro-Wilk Statistic:", round(shapiro_stat, 4))
print("Shapiro-Wilk p-value:", round(shapiro_p, 4))

if shapiro_p > 0.05:
    print("Result: Normality is not rejected at 5% significance.")
else:
    print("Result: Evidence suggests the scores are not normally distributed.")

# ------------------------------------------------------------
# 8. DISCRETE PROBABILITY DISTRIBUTION
# Placement Status: Bernoulli Distribution
# ------------------------------------------------------------

print("\n8. DISCRETE PROBABILITY DISTRIBUTION")

print("P(Placed = 1):", round(p_placed, 4))
print("P(Not Placed = 0):", round(p_not_placed, 4))

plt.figure(figsize=(7, 4))
plt.bar(
    ["Not Placed (0)", "Placed (1)"],
    [p_not_placed, p_placed]
)
plt.title("Placement Status Probability Distribution")
plt.ylabel("Probability")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 9. DISTRIBUTION OF SAMPLE MEAN
# Central Limit Theorem Demonstration
# ------------------------------------------------------------

print("\n9. DISTRIBUTION OF SAMPLE MEAN")

population = exam_scores.to_numpy()
sample_size = min(30, len(population))
number_of_samples = 1000

rng = np.random.default_rng(42)

sample_means = [
    rng.choice(population, size=sample_size, replace=True).mean()
    for _ in range(number_of_samples)
]

plt.figure(figsize=(9, 5))
sns.histplot(sample_means, bins=30, kde=True)
plt.axvline(
    np.mean(sample_means),
    linestyle="--",
    label="Mean of Sample Means"
)
plt.title("Sampling Distribution of Exam Score Mean")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()

print("Sample Size:", sample_size)
print("Number of Samples:", number_of_samples)
print("Mean of Sample Means:", round(np.mean(sample_means), 3))
print("Original Population Mean:", round(np.mean(population), 3))
print("Standard Error of Sample Means:",
      round(np.std(sample_means, ddof=1), 3))

# ------------------------------------------------------------
# 10. NAIVE BAYES THEOREM: PROBABILITY DEMONSTRATION
# ------------------------------------------------------------

print("\n10. BAYES THEOREM")

# Illustrative example, not measured medical or student data
prior = 0.30
likelihood = 0.80
false_positive_probability = 0.20

evidence_probability = (
    likelihood * prior
    + false_positive_probability * (1 - prior)
)

posterior = (likelihood * prior) / evidence_probability

print("Prior Probability:", prior)
print("Likelihood:", likelihood)
print("Evidence Probability:", round(evidence_probability, 4))
print("Posterior Probability:", round(posterior, 4))

# ------------------------------------------------------------
# 11. EIGENVALUES AND EIGENVECTORS
# Correlation Matrix Analysis
# ------------------------------------------------------------

print("\n11. EIGENVALUES AND EIGENVECTORS")

eigen_matrix = correlation_matrix.to_numpy()

eigenvalues, eigenvectors = np.linalg.eigh(eigen_matrix)

order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

print("Eigenvalues:")
print(np.round(eigenvalues, 4))

print("\nEigenvectors:")
print(np.round(eigenvectors, 4))

explained_variance_ratio = eigenvalues / eigenvalues.sum()

print("\nExplained Variance Ratio:")
print(np.round(explained_variance_ratio, 4))

# ------------------------------------------------------------
# 12. BASIC CALCULUS / DERIVATIVE DEMONSTRATION
# Linear Regression Loss Function
# ------------------------------------------------------------

print("\n12. BASIC CALCULUS")

# Simple linear regression: y = wx + b
x_values = data["study_hours"].to_numpy(dtype=float)
y_values = data["exam_score"].to_numpy(dtype=float)

w = 0.0
b = 0.0
learning_rate = 0.00001
epochs = 500

loss_history = []

for epoch in range(epochs):
    predictions = w * x_values + b
    errors = predictions - y_values

    # Derivatives of Mean Squared Error
    dw = (2 / len(x_values)) * np.sum(errors * x_values)
    db = (2 / len(x_values)) * np.sum(errors)

    w -= learning_rate * dw
    b -= learning_rate * db

    loss = np.mean(errors ** 2)
    loss_history.append(loss)

print("Learned Weight:", round(w, 4))
print("Learned Bias:", round(b, 4))
print("Final MSE:", round(loss_history[-1], 4))

plt.figure(figsize=(8, 4))
plt.plot(loss_history)
plt.title("MSE Loss During Gradient Descent")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 13. NAIVE BAYES CLASSIFIER ON STUDENT DATA
# ------------------------------------------------------------

print("\n13. NAIVE BAYES PLACEMENT CLASSIFICATION")

features = [
    "study_hours",
    "attendance",
    "sleep_hours",
    "internet_usage",
    "assignments_completed",
    "previous_score",
    "exam_score"
]

X = data[features]
y = data["placement_status"]

if y.nunique() < 2:
    print("Naive Bayes requires both placement classes.")
else:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    nb_model = GaussianNB()
    nb_model.fit(X_train_scaled, y_train)

    nb_predictions = nb_model.predict(X_test_scaled)

    nb_accuracy = accuracy_score(y_test, nb_predictions)

    print("Naive Bayes Accuracy:",
          round(nb_accuracy * 100, 2), "%")

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        nb_predictions,
        zero_division=0
    ))

# ------------------------------------------------------------
# 14. VECTOR AND MATRIX DEMONSTRATION
# ------------------------------------------------------------

print("\n14. VECTOR AND MATRIX")

student_vector = data[
    ["study_hours", "attendance", "exam_score"]
].iloc[0].to_numpy()

student_matrix = data[
    ["study_hours", "attendance", "exam_score"]
].head(5).to_numpy()

print("One Student Vector:")
print(student_vector)

print("\nFirst Five Students Matrix:")
print(student_matrix)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("STEP 3 MATHEMATICS AND STATISTICS COMPLETED")
print("=" * 65)

print("""
Topics covered:
1. Descriptive Statistics
2. Probability of Placement
3. Conditional Probability
4. Confidence Interval
5. Correlation Matrix
6. Continuous Probability Distribution
7. Discrete Probability Distribution
8. Sampling Distribution of Mean
9. Bayes Theorem
10. Eigenvalues and Eigenvectors
11. Basic Calculus and Derivatives
12. Naive Bayes Classification
13. Vectors and Matrices
""")
```

```python
# ============================================================
# STUDENT PERFORMANCE PREDICTION AND ANALYSIS SYSTEM
# STEP 5: MACHINE LEARNING AND AI
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.cluster import KMeans

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    classification_report,
    confusion_matrix,
    silhouette_score
)

# ------------------------------------------------------------
# 1. DATA VALIDATION
# ------------------------------------------------------------

print("=" * 65)
print("STEP 5: MACHINE LEARNING AND AI")
print("=" * 65)

data = df.copy()

required_columns = [
    "study_hours", "attendance", "sleep_hours",
    "internet_usage", "assignments_completed",
    "previous_score", "exam_score", "placement_status"
]

missing_columns = [c for c in required_columns if c not in data.columns]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

numeric_columns = [
    "study_hours", "attendance", "sleep_hours",
    "internet_usage", "assignments_completed",
    "previous_score", "exam_score"
]

for col in numeric_columns:
    data[col] = pd.to_numeric(data[col], errors="coerce")

if data["placement_status"].dtype == "object":
    data["placement_status"] = (
        data["placement_status"].astype(str).str.strip().str.lower()
        .map({"placed": 1, "not placed": 0})
    )

data = data.dropna(subset=required_columns)
data["placement_status"] = data["placement_status"].astype(int)

print("Dataset shape:", data.shape)
print("Data validation completed.")

# ------------------------------------------------------------
# 2. SUPERVISED LEARNING - REGRESSION
# Predict Exam Score
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("2. SUPERVISED LEARNING - EXAM SCORE REGRESSION")
print("=" * 65)

regression_features = [
    "study_hours",
    "attendance",
    "sleep_hours",
    "internet_usage",
    "assignments_completed",
    "previous_score"
]

X_reg = data[regression_features]
y_reg = data["exam_score"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)

# Model 1: Linear Regression
linear_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

linear_model.fit(X_train_reg, y_train_reg)
linear_predictions = linear_model.predict(X_test_reg)

# Model 2: Random Forest Regression
random_forest_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

random_forest_model.fit(X_train_reg, y_train_reg)
rf_predictions = random_forest_model.predict(X_test_reg)

# Evaluation
linear_mae = mean_absolute_error(y_test_reg, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test_reg, linear_predictions))
linear_r2 = r2_score(y_test_reg, linear_predictions)

rf_mae = mean_absolute_error(y_test_reg, rf_predictions)
rf_rmse = np.sqrt(mean_squared_error(y_test_reg, rf_predictions))
rf_r2 = r2_score(y_test_reg, rf_predictions)

regression_results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [linear_mae, rf_mae],
    "RMSE": [linear_rmse, rf_rmse],
    "R2 Score": [linear_r2, rf_r2]
})

print("\nRegression Model Comparison:")
print(regression_results.round(3))

# Actual vs Predicted
comparison = pd.DataFrame({
    "Actual Exam Score": y_test_reg.iloc[:10].values,
    "Linear Regression": linear_predictions[:10],
    "Random Forest": rf_predictions[:10]
})

print("\nActual vs Predicted Scores:")
print(comparison.round(2))

# Plot predictions
plt.figure(figsize=(8, 5))
plt.scatter(y_test_reg, rf_predictions, alpha=0.5)
plt.xlabel("Actual Exam Score")
plt.ylabel("Predicted Exam Score")
plt.title("Random Forest: Actual vs Predicted Exam Scores")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 3. SUPERVISED LEARNING - CLASSIFICATION
# Predict Placement Status
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("3. SUPERVISED LEARNING - PLACEMENT CLASSIFICATION")
print("=" * 65)

# Exam score is intentionally excluded because this model
# predicts placement without requiring the final exam score.

classification_features = [
    "study_hours",
    "attendance",
    "sleep_hours",
    "internet_usage",
    "assignments_completed",
    "previous_score"
]

X_class = data[classification_features]
y_class = data["placement_status"]

if y_class.nunique() != 2:
    raise ValueError("Placement classification requires two classes.")

X_train_class, X_test_class, y_train_class, y_test_class = (
    train_test_split(
        X_class,
        y_class,
        test_size=0.20,
        random_state=42,
        stratify=y_class
    )
)

logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

logistic_model.fit(X_train_class, y_train_class)

class_predictions = logistic_model.predict(X_test_class)

class_accuracy = accuracy_score(
    y_test_class,
    class_predictions
)

print("Logistic Regression Accuracy:",
      round(class_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(
    y_test_class,
    class_predictions,
    zero_division=0
))

# Confusion Matrix
cm = confusion_matrix(y_test_class, class_predictions)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not Placed", "Placed"],
    yticklabels=["Not Placed", "Placed"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Placement Classification Confusion Matrix")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 4. UNSUPERVISED LEARNING - K-MEANS CLUSTERING
# Group Students by Performance Patterns
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("4. UNSUPERVISED LEARNING - K-MEANS CLUSTERING")
print("=" * 65)

clustering_features = [
    "study_hours",
    "attendance",
    "sleep_hours",
    "internet_usage",
    "assignments_completed",
    "previous_score",
    "exam_score"
]

X_cluster = data[clustering_features]

cluster_scaler = StandardScaler()
X_cluster_scaled = cluster_scaler.fit_transform(X_cluster)

# Three student groups
kmeans_model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

data["student_cluster"] = kmeans_model.fit_predict(
    X_cluster_scaled
)

silhouette = silhouette_score(
    X_cluster_scaled,
    data["student_cluster"]
)

print("Number of Clusters:", 3)
print("Silhouette Score:", round(silhouette, 4))

# Cluster profiles
cluster_summary = data.groupby("student_cluster")[
    clustering_features
].mean().round(2)

print("\nStudent Cluster Profiles:")
print(cluster_summary)

print("\nStudents in Each Cluster:")
print(data["student_cluster"].value_counts().sort_index())

# Cluster visualization
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=data,
    x="study_hours",
    y="exam_score",
    hue="student_cluster",
    palette="Set1",
    alpha=0.6
)

plt.title("K-Means Student Performance Clusters")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. MODEL COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("5. FINAL MODEL COMPARISON")
print("=" * 65)

print("\nRegression Results:")
print(regression_results.round(3))

print("\nClassification Result:")
print("Logistic Regression Accuracy:",
      round(class_accuracy * 100, 2), "%")

print("\nClustering Result:")
print("K-Means Silhouette Score:",
      round(silhouette, 4))

# ------------------------------------------------------------
# 6. SAVE TRAINED MODELS
# ------------------------------------------------------------

joblib.dump(linear_model, "linear_regression_model.pkl")
joblib.dump(random_forest_model, "random_forest_model.pkl")
joblib.dump(logistic_model, "logistic_placement_model.pkl")
joblib.dump(kmeans_model, "kmeans_student_model.pkl")
joblib.dump(cluster_scaler, "kmeans_scaler.pkl")

print("\nAll trained models saved successfully.")

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("STEP 5: MACHINE LEARNING AND AI COMPLETED")
print("=" * 65)

print("""
TOPICS COMPLETED:

1. Machine Learning using Scikit-learn
2. Supervised Learning
3. Linear Regression
4. Random Forest Regression
5. Logistic Regression Classification
6. Unsupervised Learning
7. K-Means Clustering
8. Model Evaluation and Comparison
9. Model Saving

Not implemented:
- LLM: Not required for this prediction project.
- Reinforcement Learning: Not suitable for this static dataset.
""")
```

```python
# STEP 8: DEEP LEARNING USING PYTORCH
# Artificial Neural Network (ANN)

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1. Select input and target
features = [
    "study_hours",
    "attendance",
    "sleep_hours",
    "internet_usage",
    "assignments_completed",
    "previous_score"
]

X = df[features].values.astype(np.float32)
y = df["exam_score"].values.astype(np.float32).reshape(-1, 1)

# 2. Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 3. Normalize data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 4. Convert data into PyTorch tensors
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32)

# 5. Create ANN Model
class StudentANN(nn.Module):
    def __init__(self):
        super(StudentANN, self).__init__()

        self.network = nn.Sequential(
            nn.Linear(6, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.network(x)

model = StudentANN()

# 6. Define loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 7. Train ANN Model
epochs = 100
loss_history = []

for epoch in range(epochs):

    model.train()

    predictions = model(X_train_tensor)
    loss = criterion(predictions, y_train_tensor)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    loss_history.append(loss.item())

    if (epoch + 1) % 20 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")

# 8. Evaluate Model
model.eval()

with torch.no_grad():
    test_predictions = model(X_test_tensor).numpy()

print("\nANN MODEL PERFORMANCE")
print("MAE:", mean_absolute_error(y_test, test_predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_test, test_predictions)))
print("R2 Score:", r2_score(y_test, test_predictions))

# 9. Plot Training Loss
plt.figure(figsize=(8, 5))
plt.plot(loss_history)
plt.title("ANN Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()

# 10. Compare Actual and Predicted Scores
comparison = pd.DataFrame({
    "Actual Score": y_test.flatten()[:10],
    "Predicted Score": test_predictions.flatten()[:10]
})

display(comparison)

# 11. Save ANN model
torch.save(model.state_dict(), "student_ann_model.pth")

print("\nDEEP LEARNING COMPLETED!")
print("ANN Model Saved Successfully!")
```
